Fibonacci Series

```python
%%writefile Fibonacci.c
#include <stdio.h>

int fib_recursive(int n) {
    if (n <= 1) return n;
    return fib_recursive(n - 1) + fib_recursive(n - 2);
}

void fib_iterative(int n) {
    int a = 0, b = 1, c;
    for (int i = 0; i < n; i++) {
        printf("%d ", a);
        c = a + b;
        a = b;
        b = c;
    }
}

int main() {
    int n;
    Print("Enter the number of terms: ");
    scanf("%d", &n);

    printf("Iterative: ");
    fib_iterative(n);

    printf("\nRecursive: ");
    for (int i = 0; i < n; i++)
        printf("%d ", fib_recursive(i));

    return 0;
}
```
