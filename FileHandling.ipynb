#### Program for File handling:

```python
%%writefile file_handling.c
#include <stdio.h>

int main() {
    FILE *fp;
    char text[100];

    fp = fopen("sample.txt", "w");
    printf("Enter text to write: ");

    fgets(text, 100, stdin);
    fputs(text, fp);
    fclose(fp);

    fp = fopen("sample.txt", "r");
    printf("\nFile content:\n");

    while (fgets(text, 100, fp)) {
        printf("%s", text);
    }

    fclose(fp);

    return 0;
}
```

```python
!gcc file_handling.c -o file_handling.exe
```

```python
!./file_handling.exe
```

#### Program for Apending a Text:

```python
%%writefile apend.c
#include <stdio.h>
int main() {
    FILE *fp;
    char text[100];
    fp = fopen("sample.txt", "a");
    printf("Enter text to append: ");
    fgets(text, sizeof(text), stdin);
    fputs(text, fp);
    fclose(fp);
    printf("Text appended successfully.");
    return 0;
}
```

```python
!gcc apend.c -o apend.exe
```

```python
!./apend.exe
```

#### Program for Counting characters

```python
%%writefile chars.c
#include <stdio.h>

int main() {
    FILE *fp;
    char ch;
    int chars = 0, lines = 0;

    fp = fopen("sample.txt", "r");
    if (fp == NULL) {
      printf("File not found.");
      return 1;
      }

        while ((ch = fgetc(fp)) != EOF) {
        chars++;
        if (ch == '\n') lines++;
    }
    fclose(fp);
    printf("Characters: %d\nLines: %d", chars, lines);
 return 0;
}
```

#### Program for File size

```python
%%writefile close.c
#include <stdio.h>

int main() {
    FILE *fp;
    long file_size;

    fp = fopen("sample.txt", "r");
    if (fp == NULL) {
        printf("Error: Could not open file 'sample.txt'\n");
        return 1;
    }

    fseek(fp, 0, SEEK_END);

    file_size = ftell(fp);
    printf("File size: %ld bytes\n", file_size);

    rewind(fp);
    printf("File pointer reset to beginning. Current position: %ld bytes\n", ftell(fp));

    fclose(fp);
    return 0;
}
```

```python
!gcc close.c -o close.exe
```

```python
!./close.exe
```

#### Program for opening file

```python
%%writefile open.c
#include <stdio.h>

int main() {
    FILE *src, *dest;
    char ch;

    src = fopen("source.txt", "r");
    dest = fopen("copy.txt", "w");

    if (src == NULL || dest == NULL) {
        printf("Error opening file.\n");
        return 1;
    }

    while ((ch = fgetc(src)) != EOF) {
        fputc(ch, dest);
    }

    printf("File copied successfully.");
    fclose(src);
    fclose(dest);
    return 0;
}
```

```python
!gcc open.c -o open.exe
```
