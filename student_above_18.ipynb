```python
%%writefile student.c
#include <stdio.h>

struct student {
    char name[50];
    int roll;
    int age;
};

int main() {
    struct student s[10];
    int count = 0;

    for (int i = 0; i < 10; i++) {
        printf("Enter name, roll, age of student %d: ", i + 1);
        scanf("%s%d%d", s[i].name, &s[i].roll, &s[i].age);
        if (s[i].age > 18) count++;
    }

    printf("\nNumber of students above 18: %d", count);
    return 0;
}
```
